LOAD LIBRARIES

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

LOAD DATA

```python
customers = pd.read_csv('customers.csv')
geolocation = pd.read_csv('geolocation.csv')
orders = pd.read_csv('orders.csv')
order_items = pd.read_csv('order_items.csv')
order_payments = pd.read_csv('order_payments.csv')
order_reviews = pd.read_csv('order_reviews.csv')
products = pd.read_csv('products.csv')
sellers = pd.read_csv('sellers.csv')
```

```python
orders.drop(columns=['order_delivered_carrier_date','order_delivered_customer_date'], inplace=True)
```

MERGING

```python
all = orders.merge(order_items, on='order_id').merge(order_payments, on='order_id').merge(order_reviews, on='order_id').merge(products, on='product_id').merge(sellers, on='seller_id').merge(customers, on='customer_id').merge(geolocation, left_on='customer_zip_code_prefix', right_on='zip_code_prefix')
```

```python
all.drop(columns=['customer_zip_code_prefix','zip_code_prefix', 'order_id', 'customer_id'], inplace=True)
```

```python
all.drop(columns=['product_id', 'seller_id', 'customer_unique_id', 'seller_zip_code_prefix', ], inplace=True)
```

```python
all.drop(columns=['seller_city', 'seller_state', 'seller_contact_age', 'seller_contact_gender', 'review_id'], inplace=True)
```

```python
all.drop(columns=['order_item_id', 'payment_value', 'seller_company_name'], inplace=True)
```
